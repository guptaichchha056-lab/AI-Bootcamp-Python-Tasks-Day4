```python
# Import necessary libraries
import collections
from collections import Counter
import matplotlib.pyplot as plt
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download("punkt_tab")
nltk.download('stopwords')
```

```python
# Provided text
text = """
One of my main hobbies and interests is exploring the world of technology, especially coding and software development. I enjoy learning new programming concepts and applying them through practical problems and projects. I am currently pursuing B.Tech in Computer Science Engineering, and I regularly practice programming to improve my problem-solving and logical thinking skills. I have been learning Data Structures and Algorithms (DSA) and solving coding problems using C++, which has helped me understand concepts such as arrays, strings, functions, recursion, searching, sorting, hashing, and basic object-oriented programming. Along with DSA, I am also exploring Web Development and learning how frontend and backend technologies work together to build functional websites. I have worked with HTML and CSS and am gradually learning JavaScript, Node.js, Express.js, and databases to understand how complete web applications are developed. I also enjoy participating in hackathons and technical events because they give me an opportunity to work on real-world problems, collaborate with teammates, and turn ideas into practical solutions. I have worked on projects such as a Hospital Management System using C++ and have also explored web-based ideas such as a Campus Event Management System. I am also developing my understanding of Python and Machine Learning by practicing NumPy, Pandas, Matplotlib, data cleaning, regression, decision trees, and basic model evaluation. Apart from coding, I enjoy exploring new technologies, attending technical events, and continuously improving my communication and technical skills. For me, coding is not just an academic subject but a hobby that allows me to learn, experiment, solve problems, and create useful solutions.
"""
print(text)
```

```python
# Tokenization of words
words = word_tokenize(text.lower())  # Convert to lowercase for uniformity
print(words)
```

```python
# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
#Frequency distribution of words
word_freq = Counter(filtered_words) 
print(word_freq)

# Create a WordCloud
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
# Plotting the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.savefig(
    "wordcloud.jpg", format="jpg", dpi=300, bbox_inches="tight", pad_inches=0
)
plt.show()
```
