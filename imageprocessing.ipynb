```python
import cv2 #used for computer vision 
import numpy as np
import matplotlib.pyplot as plt
```

```python
#to read image
img=cv2.imread(r"C:\Users\WIN\ichchha.jpg") #for url-copy as path and keep iamge in same folder as all ipynb files
print(type(img))
```

```python
#to print image on screen
cv2.imshow("Photos",img)
```

```python
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
print(img.shape) #shape is property dimension is coming 950 and 822 ...3 is color sequence
```

```python
#resize imagee
img_resize=cv2.resize(img,(400,256))
cv2.imshow("Photos:,img_resize)
cv2.waitkey(0)
cv2.destroyAllWindows()
```

```python
print(img_resize.shape)
```

```python
#flip image
img_flip = cv2.flip(img,1)
```

```python
cv2.imshow("flipped Image",img_flip)
cv2.waitkey(0)
cv2.destroyAllWindows()
```

```python
#image crop
img_crop=img[100:300, 200:500]
cv2.imshow("Cropped Image",img_crop)
cv2.waitkey(0)
cv2.destroyAllWindows()
```
